````python
import json 
import re


file = "./selected_emotional_cherries_metadata.jsonl"

with open(file, 'r') as f:
    data = [json.loads(line) for line in f]

# alphabetical sort based on key w/out 'low' or 'high'
sorted_data = sorted(data, key=lambda x: x['key'].replace('low','').replace('high',''))

# overwrite
sorted_data = data
````

````python
def sentence_case(text):
    if not text:
        return text
    
    # normalise whitespace around punctuation
    text = re.sub(r'\s*([.!?])\s*', r'\1 ', text).strip()
    
    # punctuations -> split sentences
    sentences = [s.strip() for s in re.split(r'(?<=[.!?])\s+', text) if s.strip()]
    
    processed = []
    for sent in sentences:
        if not sent:
            continue
        
        # sentence case
        if sent:
            sent = sent[0].upper() + sent[1:]
        
        # standalone 'i' -> 'I'
        # Match ' i ' or start/end boundaries
        sent = re.sub(r'\bi\b', 'I', sent)
        
        processed.append(sent)
    
    output_text = ' '.join(processed)

    if output_text[-1] not in ["?", "!", "."]:
        output_text = output_text + "."
    return output_text

# Test
# test_cases = [
#     "hello world. i am here. what about this? i think it works.",
#     "i went to the store. it was closed. i came home.",
#     "multiple   spaces.   i  am  testing.",
#     ""
# ]

# for t in test_cases:
#     print(f"Input:  '{t}'")
#     print(f"Output: '{sentence_case(t)}'")
#     print("-" * 40)
````

````python
titles, texts = [], []
for i, d in enumerate(sorted_data):
    # sentence case 
    print(d)
    title = f"{i+1}) {d['emotion'].capitalize()}"
    text = sentence_case(d['text'])
    
    titles.append(title)
    texts.append(text)


with open("title.json", 'w') as f:
    json.dump(titles, f, indent=4, ensure_ascii=False)

with open("info.json", 'w') as f:
    json.dump(texts, f, indent=4, ensure_ascii=False)
````
